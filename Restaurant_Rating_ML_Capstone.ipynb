# **Project Summary -**

The Zomato Restaurant Clustering project aims to analyze restaurant and customer review data using Unsupervised Machine Learning techniques. Zomato is one of India's largest online food delivery and restaurant discovery platforms, containing information about thousands of restaurants, cuisines, customer ratings, reviews, and pricing. With such a large volume of data, manually identifying patterns among restaurants is challenging. This project addresses that challenge by grouping similar restaurants based on their characteristics and customer feedback.

The project utilizes two datasets: one containing restaurant metadata, including restaurant names, cuisines, collections, timings, and estimated cost, and another containing customer reviews, ratings, reviewer metadata, and review timestamps. These datasets are cleaned, preprocessed, and analyzed to gain meaningful business insights.

Exploratory Data Analysis (EDA) is performed to understand customer preferences, restaurant popularity, pricing distribution, cuisine trends, and rating patterns. Various visualizations such as histograms, bar charts, box plots, heatmaps, scatter plots, and pair plots are created to identify relationships between different variables.

Before building the machine learning model, the datasets undergo preprocessing steps including handling missing values, removing duplicates, converting categorical variables into machine-readable formats, feature engineering, and text preprocessing where necessary. Numerical features are standardized to ensure fair contribution during clustering.

The primary machine learning algorithm used in this project is K-Means Clustering. The optimal number of clusters is determined using the Elbow Method and validated using the Silhouette Score. Principal Component Analysis (PCA) is applied to reduce dimensionality and visualize the resulting clusters effectively. Additional clustering techniques may also be explored for comparison.

The clustering results help identify different categories of restaurants such as premium dining, budget-friendly restaurants, highly-rated establishments, and restaurants specializing in particular cuisines. These insights can assist Zomato in improving recommendation systems, designing targeted marketing campaigns, optimizing restaurant listings, and enhancing customer experience.

Overall, this project demonstrates how Unsupervised Machine Learning can transform raw restaurant and review data into valuable business intelligence. The findings provide practical recommendations that support data-driven decision-making for restaurant discovery platforms while showcasing the application of clustering techniques in real-world business scenarios.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# ==========================================================
# DATA WRANGLING
# ==========================================================

# Create working copies of the original datasets
metadata = metadata_df.copy()
reviews = reviews_df.copy()

# ----------------------------------------------------------
# Remove duplicate records
# ----------------------------------------------------------

metadata.drop_duplicates(inplace=True)
reviews.drop_duplicates(inplace=True)

# ----------------------------------------------------------
# Clean Cost column
# Remove commas and convert values to numeric
# ----------------------------------------------------------

metadata["Cost"] = (
    metadata["Cost"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

metadata["Cost"] = pd.to_numeric(
    metadata["Cost"],
    errors="coerce"
)

# ----------------------------------------------------------
# Clean Rating column
# Convert rating values to numeric
# ----------------------------------------------------------

reviews["Rating"] = (
    reviews["Rating"]
    .astype(str)
    .str.strip()
)

reviews["Rating"] = pd.to_numeric(
    reviews["Rating"],
    errors="coerce"
)

# ----------------------------------------------------------
# Remove reviews where Rating is missing
# ----------------------------------------------------------

reviews = reviews.dropna(subset=["Rating"])

# ----------------------------------------------------------
# Display cleaned data information
# ----------------------------------------------------------

print("=" * 60)
print("DATA WRANGLING COMPLETED")
print("=" * 60)

print("Metadata Shape :", metadata.shape)
print("Reviews Shape  :", reviews.shape)

print("\nCost Data Type   :", metadata["Cost"].dtype)
print("Rating Data Type :", reviews["Rating"].dtype)
````

### What all manipulations have you done and insights you found?

During the data wrangling process, the datasets were cleaned and prepared for further analysis.

The following preprocessing steps were performed:

• Created copies of the original datasets to preserve raw data.
• Removed duplicate records to improve data quality.
• Checked for missing values in both datasets.
• Filled missing values with "Unknown" where appropriate.
• Converted the Cost column into numeric format by removing commas.
• Converted the Rating column into numeric values.
• Removed rows having missing ratings because ratings are essential for customer analysis.

These preprocessing steps improved the quality of the datasets and made them suitable for exploratory data analysis and machine learning. The cleaned datasets are now consistent and ready for feature engineering, visualization, and clustering.

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
There is no statistically significant difference in the average customer ratings between reviews with pictures and reviews without pictures.

Alternative Hypothesis (H₁):
There is a statistically significant difference in the average customer ratings between reviews with pictures and reviews without pictures.

#### 2. Perform an appropriate statistical test.

````python
# ==========================================================
# Hypothesis Testing - Independent Two-Sample T-Test
# ==========================================================

from scipy.stats import ttest_ind
import pandas as pd

# Convert Rating to numeric
# Invalid values will be converted to NaN
reviews_df["Rating"] = pd.to_numeric(
    reviews_df["Rating"],
    errors="coerce"
)

# Convert Pictures to numeric
reviews_df["Pictures"] = pd.to_numeric(
    reviews_df["Pictures"],
    errors="coerce"
)

# Remove rows where Rating or Pictures is missing
test_data = reviews_df.dropna(
    subset=["Rating", "Pictures"]
)

# Group 1: Reviews with pictures
ratings_with_pictures = test_data.loc[
    test_data["Pictures"] > 0,
    "Rating"
]

# Group 2: Reviews without pictures
ratings_without_pictures = test_data.loc[
    test_data["Pictures"] == 0,
    "Rating"
]

# Perform Independent Two-Sample T-Test
t_statistic, p_value = ttest_ind(
    ratings_with_pictures,
    ratings_without_pictures,
    equal_var=False
)

# Display results
print("=" * 60)
print("HYPOTHESIS TESTING - PICTURES VS CUSTOMER RATING")
print("=" * 60)

print(
    f"Average Rating - With Pictures    : "
    f"{ratings_with_pictures.mean():.2f}"
)

print(
    f"Average Rating - Without Pictures : "
    f"{ratings_without_pictures.mean():.2f}"
)

print(f"T-Statistic : {t_statistic:.4f}")
print(f"P-Value     : {p_value:.4f}")

# Decision using 5% significance level
alpha = 0.05

print("\nSignificance Level (α):", alpha)

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis.")
    print(
        "Conclusion: There is a statistically significant "
        "difference between the two groups."
    )
else:
    print("Decision: Fail to Reject the Null Hypothesis.")
    print(
        "Conclusion: There is no statistically significant "
        "difference between the two groups."
    )
````

##### Which statistical test have you done to obtain P-Value?

The obtained P-value is 0.0000, which is less than the significance level of 0.05. Therefore, the Null Hypothesis is rejected.

This indicates that there is a statistically significant difference in the average customer ratings between reviews with pictures and reviews without pictures. Reviews with pictures have a higher average rating of 3.86 compared with 3.54 for reviews without pictures.

##### Why did you choose the specific statistical test?

The result provides a useful business insight for Zomato. Reviews containing pictures are associated with higher customer ratings in this dataset. Zomato can encourage customers to add pictures to their reviews because visual reviews can provide richer information to other customers.

However, statistical significance does not mean that pictures directly cause higher ratings. Other factors such as food quality, service, pricing, and customer experience may also influence ratings.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
There is no statistically significant difference in the average customer ratings among different cuisines.

Alternative Hypothesis (H₁):
There is a statistically significant difference in the average customer ratings among different cuisines.

#### 2. Perform an appropriate statistical test.

````python
# ==========================================================
# Hypothetical Statement - 3
# Cuisine vs Customer Rating
# ==========================================================

import pandas as pd
from scipy.stats import f_oneway

# Prepare restaurant metadata
cuisine_data = metadata_df[["Name", "Cuisines"]].copy()

# Prepare reviews
rating_data = reviews_df[["Restaurant", "Rating"]].copy()

# Convert Rating to numeric
rating_data["Rating"] = pd.to_numeric(
    rating_data["Rating"],
    errors="coerce"
)

# Remove missing ratings
rating_data = rating_data.dropna(subset=["Rating"])

# Merge restaurant information with ratings
merged_cuisine_data = pd.merge(
    rating_data,
    cuisine_data,
    left_on="Restaurant",
    right_on="Name",
    how="inner"
)

# Remove missing cuisines
merged_cuisine_data = merged_cuisine_data.dropna(
    subset=["Cuisines"]
)

# Take the first cuisine for restaurants having multiple cuisines
merged_cuisine_data["Main_Cuisine"] = (
    merged_cuisine_data["Cuisines"]
    .astype(str)
    .str.split(",")
    .str[0]
    .str.strip()
)

# Select the 5 most common cuisines
top_cuisines = (
    merged_cuisine_data["Main_Cuisine"]
    .value_counts()
    .head(5)
    .index
)

# Create rating groups
rating_groups = [
    merged_cuisine_data.loc[
        merged_cuisine_data["Main_Cuisine"] == cuisine,
        "Rating"
    ]
    for cuisine in top_cuisines
]

# Perform One-Way ANOVA
f_statistic, p_value = f_oneway(*rating_groups)

# Display results
print("=" * 60)
print("HYPOTHESIS TESTING - CUISINE VS CUSTOMER RATING")
print("=" * 60)

print("\nCuisines tested:")

for cuisine, ratings in zip(top_cuisines, rating_groups):
    print(
        f"{cuisine}: {len(ratings)} reviews, "
        f"Average Rating = {ratings.mean():.2f}"
    )

print(f"\nF-Statistic : {f_statistic:.4f}")
print(f"P-Value     : {p_value:.4f}")

alpha = 0.05

print("\nSignificance Level (α):", alpha)

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis.")
    print(
        "Conclusion: There is a statistically significant "
        "difference in average ratings among the cuisine groups."
    )
else:
    print("Decision: Fail to Reject the Null Hypothesis.")
    print(
        "Conclusion: There is no statistically significant "
        "difference in average ratings among the cuisine groups."
    )
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA (Analysis of Variance)

#### 1. Expand Contraction

````python
# Expand Contractions

import re

def expand_contractions(text):
    contractions = {
        "can't": "cannot",
        "won't": "will not",
        "n't": " not",
        "'re": " are",
        "'s": " is",
        "'d": " would",
        "'ll": " will",
        "'t": " not",
        "'ve": " have",
        "'m": " am"
    }

    text = str(text)

    for contraction, expansion in contractions.items():
        text = re.sub(contraction, expansion, text, flags=re.IGNORECASE)

    return text

test_data['Review'] = test_data['Review'].apply(expand_contractions)

print("Contraction expansion completed.")
````

````python
# Text Vectorization using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(max_features=5000)

X_tfidf = tfidf_vectorizer.fit_transform(
    test_data['Review'].fillna('')
)

print("TF-IDF vectorization completed.")
print("TF-IDF matrix shape:", X_tfidf.shape)
````

##### Which text vectorization technique have you used and why?

I used TF-IDF (Term Frequency-Inverse Document Frequency) for text vectorization. It converts the review text into numerical features by giving higher importance to words that are relevant to particular reviews and lower importance to commonly occurring words. I used TF-IDF because it is suitable for converting textual review data into numerical features for machine learning models.

### 6. Data Scaling

````python
# Data Scaling using StandardScaler

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Scale the numerical features
numerical_columns = ['Rating', 'Pictures', 'Pictures_Log']

test_data[numerical_columns] = scaler.fit_transform(
    test_data[numerical_columns]
)

print("Data scaling completed.")
print(test_data[numerical_columns].head())
````

##### Which method have you used to scale you data and why?

### 7. Dimesionality Reduction

##### Do you think that dimensionality reduction is needed? Explain Why?

Yes, dimensionality reduction is needed because the TF-IDF representation contains 5,000 features. A large number of features can increase computational complexity and make model training more difficult. Reducing the feature space helps simplify the data and improve computational efficiency.

````python
# Dimensionality Reduction using PCA

from sklearn.decomposition import PCA

# Apply PCA to the TF-IDF feature matrix
pca = PCA(n_components=100, random_state=42)

X_pca = pca.fit_transform(X_tfidf.toarray())

print("Dimensionality reduction completed.")
print("Original TF-IDF shape:", X_tfidf.shape)
print("Reduced shape:", X_pca.shape)
print("Explained variance ratio:", pca.explained_variance_ratio_.sum())
````

##### Which dimensionality reduction technique have you used and why? (If dimensionality reduction done on dataset.)

I used Principal Component Analysis (PCA) to reduce the 5,000 TF-IDF features to 100 components. PCA was chosen to reduce the number of dimensions while retaining as much of the variation in the data as possible.

### 8. Data Splitting

````python
# Split data into training and testing sets

from sklearn.model_selection import train_test_split

# Features and target
X = X_pca
y = test_data['Rating']

# 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Data splitting completed.")
print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
````

##### What data splitting ratio have you used and why?

I used an 80:20 train-test split, where 80% of the data was used for training and 20% for testing. This provides sufficient data for the model to learn while keeping a separate portion of the data to evaluate its performance on unseen data.

## ***7. ML Model Implementation***

### ML Model - 1

````python
# ML Model - 1: Random Forest Regression

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the model
model_1 = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Fit the algorithm
model_1.fit(X_train, y_train)

# Predict on the test data
y_pred_1 = model_1.predict(X_test)

# Evaluate the model
mae_1 = mean_absolute_error(y_test, y_pred_1)
mse_1 = mean_squared_error(y_test, y_pred_1)
rmse_1 = np.sqrt(mse_1)
r2_1 = r2_score(y_test, y_pred_1)

print("Random Forest Regression Results:")
print("MAE :", mae_1)
print("MSE :", mse_1)
print("RMSE:", rmse_1)
print("R2 Score:", r2_1)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

Random Forest Regression was used as the first machine learning model to predict the Rating. The model achieved an MAE of 0.503, MSE of 0.428, RMSE of 0.654, and an R² score of 0.571. The R² score indicates that the model explains approximately 57.1% of the variation in the target variable. The evaluation chart shows the performance of the model using these metrics.

````python
# Visualizing Evaluation Metric Score Chart

import matplotlib.pyplot as plt

metrics = ['MAE', 'MSE', 'RMSE', 'R2 Score']
scores = [mae_1, mse_1, rmse_1, r2_1]

plt.figure(figsize=(8, 5))
plt.bar(metrics, scores)
plt.title('Random Forest Regression - Evaluation Metrics')
plt.xlabel('Evaluation Metrics')
plt.ylabel('Score')
plt.tight_layout()
plt.show()
````

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# Hyperparameter Tuning using GridSearchCV

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

# Smaller parameter grid for faster execution
param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [10, None]
}

grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=3,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)

mae_tuned = mean_absolute_error(y_test, y_pred_tuned)
mse_tuned = mean_squared_error(y_test, y_pred_tuned)
rmse_tuned = np.sqrt(mse_tuned)
r2_tuned = r2_score(y_test, y_pred_tuned)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nTuned Random Forest Results:")
print("MAE :", mae_tuned)
print("MSE :", mse_tuned)
print("RMSE:", rmse_tuned)
print("R2 Score:", r2_tuned)
````

##### Which hyperparameter optimization technique have you used and why?

GridSearchCV was used for hyperparameter optimization. It systematically tested different combinations of Random Forest parameters using 3-fold cross-validation to identify the best configuration. The best parameters were n_estimators = 100 and max_depth = None.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

No significant improvement was observed after hyperparameter tuning. The evaluation metrics remained the same, with MAE = 0.503, MSE = 0.428, RMSE = 0.654, and R² Score = 0.571. Therefore, the tuned model performed similarly to the original Random Forest model.
